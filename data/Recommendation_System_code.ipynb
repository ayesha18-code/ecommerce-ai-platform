Product-Product Recommendation

```python
def recommend_similar_products(product_id, top_n=5):

    if product_id not in products["product_id"].values:
        print("Product ID not found!")
        return

    product_index = products[
        products["product_id"] == product_id
    ].index[0]

    similarity_scores = list(
        enumerate(text_similarity[product_index])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similar_indices = [
        i for i, score in similarity_scores[1:top_n+1]
    ]

    result = products.iloc[similar_indices][
        [
            "product_id",
            "product_category_name",
            "product_category_name_english"
        ]
    ]

    return result
```

**Hybrid Recommendation**

```python
import numpy as np

def hybrid_recommendation(product_id, top_n=5):

    if product_id not in products["product_id"].values:
        print("Product ID not found!")
        return

    product_index = products[
        products["product_id"] == product_id
    ].index[0]

    similarity_scores = text_similarity[product_index]

    result = products.copy()

    result["similarity_score"] = similarity_scores

    # Normalize rating
    max_rating = result["average_rating"].max()

    if max_rating > 0:
        result["rating_score"] = (
            result["average_rating"] / max_rating
        )
    else:
        result["rating_score"] = 0

    # Normalize popularity
    max_popularity = result["purchase_count"].max()

    if max_popularity > 0:
        result["popularity_score"] = (
            result["purchase_count"] / max_popularity
        )
    else:
        result["popularity_score"] = 0

    # Hybrid score
    result["hybrid_score"] = (
        0.6 * result["similarity_score"]
        + 0.2 * result["rating_score"]
        + 0.2 * result["popularity_score"]
    )

    # Remove selected product
    result = result[
        result["product_id"] != product_id
    ]

    result = result.sort_values(
        "hybrid_score",
        ascending=False
    )

    return result[
        [
            "product_id",
            "product_category_name_english",
            "average_rating",
            "purchase_count",
            "hybrid_score"
        ]
    ].head(top_n)
```

Final Product Score

```python
def hybrid_product_recommendation(product_id, top_n=5):

    if product_id not in products["product_id"].values:
        print("Product ID not found!")
        return pd.DataFrame()

    product_index = products[
        products["product_id"] == product_id
    ].index[0]

    result = products.copy()

    # Similarity
    result["similarity_score"] = text_similarity[
        product_index
    ]

    # Review score
    max_rating = result["average_rating"].max()

    if max_rating > 0:
        result["review_score"] = (
            result["average_rating"] / max_rating
        )
    else:
        result["review_score"] = 0

    # Popularity score
    max_purchase = result["purchase_count"].max()

    if max_purchase > 0:
        result["popularity_score"] = (
            result["purchase_count"] / max_purchase
        )
    else:
        result["popularity_score"] = 0

    # Final Score
    result["final_product_score"] = (
        0.6 * result["similarity_score"]
        + 0.2 * result["review_score"]
        + 0.2 * result["popularity_score"]
    )

    # Remove selected product
    result = result[
        result["product_id"] != product_id
    ]

    result = result.sort_values(
        "final_product_score",
        ascending=False
    )

    return result[
        [
            "product_id",
            "product_category_name_english",
            "similarity_score",
            "review_score",
            "popularity_score",
            "final_product_score"
        ]
    ].head(top_n)
```
